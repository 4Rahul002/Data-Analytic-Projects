###Impute missing values

````python
# 1. Customer Remarks - Impute with "No Feedback" if missing
df['Customer Remarks'].fillna('No Feedback', inplace=True)
# 2. Order Date & Item Price - Impute missing Item_price by Product Category Median
df['Item_price'] = df.groupby('Product_category')['Item_price'].transform(lambda x: x.fillna(x.median()))
# 3. Connected Handling Time - Impute using mean by Agent Shift
df['connected_handling_time'] = df.groupby('Agent Shift')['connected_handling_time'].transform(lambda x: x.fillna(x.mean()))
# 4. Customer City & Product Category - Impute with mode within similar CSAT Score
df['Customer_City'] = df.groupby('CSAT Score')['Customer_City'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Unknown'))
df['Product_category'] = df.groupby('CSAT Score')['Product_category'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Unknown'))

# 5. Issue Resolution Time - Impute by average time by Category & Channel
df['issue_reported_at'] = pd.to_datetime(df['Issue_reported at'], errors='coerce')
df['issue_responded'] = pd.to_datetime(df['issue_responded'], errors='coerce')
df['resolution_time'] = (df['issue_responded'] - df['issue_reported_at']).dt.total_seconds() / 3600

df['resolution_time'] = df.groupby(['category', 'channel_name'])['resolution_time'].transform(lambda x: x.fillna(x.mean()))
````

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
#Check Any Outliers
plt.figure(figsize=(10,6))
sns.boxplot(df)
plt.show()
````

````python
# Function to calculate outliers using IQR
def calculate_outliers(df, column):
    Q1 = df[column].quantile(0.25)  # First Quartile
    Q3 = df[column].quantile(0.75)  # Third Quartile
    IQR = Q3 - Q1                   # Interquartile Range

    # Outlier boundaries
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify outliers
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

    print(f"Outlier Summary for '{column}':")
    print(f"Lower Bound: {lower_bound}, Upper Bound: {upper_bound}")
    print(f"Total Outliers: {len(outliers)}")

    return outliers

# Function to cap (winsorize) outliers using IQR
def cap_outliers(df, column):
    Q1 = df[column].quantile(0.25)  # First Quartile
    Q3 = df[column].quantile(0.75)  # Third Quartile
    IQR = Q3 - Q1                   # Interquartile Range

    # Outlier boundaries
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Cap outliers to the boundaries
    df[column] = df[column].clip(lower_bound, upper_bound)

    print(f"Outliers in '{column}' have been capped between {lower_bound} and {upper_bound}.")
    return df

# Apply capping to all numerical columns
numerical_columns = df.select_dtypes(include='number').columns

for column in numerical_columns:
    df = cap_outliers(df, column)

# Verify changes across all numerical columns
print(df.describe())

# Calculate Outliers
outliers = calculate_outliers(df,'Item_price')
# Save the cleaned dataset
df.to_csv('cleaned_customer_support_data.csv', index=False)
print("Cleaned dataset saved successfully.")
````

##### 2. What is/are the insight(s) found from the chart?

1. Correlation Between Variables
Positive Correlation: Variables increase together (e.g., higher handling time may lead to lower CSAT scores).
Negative Correlation: As one variable increases, the other decreases (e.g., higher item price may reduce satisfaction).
No Correlation: Scattered points with no visible trend indicate little or no relationship.

 Example Insight:
If connected_handling_time shows a negative trend with CSAT Score, longer handling times reduce customer satisfaction.

 2. Clusters of CSAT Scores
Different CSAT levels may form distinct clusters across certain numeric features.
Tight Clusters: Indicates consistency (e.g., mid-range item prices might always lead to high CSAT).
Scattered Points: Suggests variability or no significant influence from that feature.

 Example Insight:
If low CSAT scores form a cluster in higher handling times, reducing handling time could improve satisfaction.

3. Detecting Outliers
Outliers appear as points far from the main distribution.
Outliers can identify problematic cases, such as extremely high handling times or unusually low satisfaction.

Example Insight:
If a small cluster of CSAT = 1 corresponds to extreme resolution times, resolving these cases faster could raise satisfaction.

 4. Multicollinearity (Feature Redundancy)
Strong correlations between two independent variables indicate redundancy.
Multicollinearity can reduce the effectiveness of predictive models.

 Example Insight:
If item_price and order_value are highly correlated, you may only need one for model training.

 5. Trends by Shift or Tenure
Agent Tenure: Longer-tenure agents may handle queries more efficiently.
Agent Shift: Certain shifts (e.g., Split) may show unique performance patterns.

Example Insight:
If connected_handling_time is lower during the morning shift and CSAT is higher, you might allocate complex tasks to that time.
