## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
numeric_df=df.select_dtypes(include='number')
plt.figure(figsize=(15,10))
sns.boxplot(data=numeric_df)
plt.xticks(rotation=90)
plt.show()
````

````python
# Outliers in duration column only
df['duration_sec'].value_counts()
````

````python
# Need to check how much data loss will occur if i use IQR method depending on that i shall decide to use winsorize method

# IQR calculation
Q1 = df['duration_sec'].quantile(0.25)
Q3 = df['duration_sec'].quantile(0.75)
IQR = Q3 - Q1

# Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter out outliers
df_iqr_cleaned = df[(df['duration_sec'] >= lower_bound) & (df['duration_sec'] <= upper_bound)]
````

````python
print(df_iqr_cleaned.shape)
````

````python
# after IQR method the outliers
numeric_df=df_iqr_cleaned.select_dtypes(include='number')
plt.figure(figsize=(15,10))
sns.boxplot(data=numeric_df)
plt.xticks(rotation=90)
plt.show()
# Now this is acceptable , so will go ahead with removing the outliers thus
# 82691 rows and 16 cols
````

#### Chart - 14 - Correlation Heatmap

````python
# Correlation Heatmap visualization code
# Select numeric columns
numeric_df = df.select_dtypes(include='number')

# Correlation matrix
corr_matrix = numeric_df.corr()

# Plot the heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
````

#### Chart - 15 - Pair Plot

````python
# Pair Plot visualization code
# Subset of columns with insights
important_cols = ['duration_sec', 'start_station_id', 'end_station_id', 'bike_id', 'user_type']

# Filter DataFrame
df_subset = df[important_cols].copy()

# Drop rows with missing values in these columns (if any)
df_subset = df_subset.dropna()

# Pairplot with user_type as hue
sns.pairplot(df_subset, hue='user_type', diag_kind='kde', plot_kws={'alpha': 0.6})
plt.suptitle("Pairplot of Important Features by User Type", y=1.02)
plt.show()
````

##### 1. Why did you pick the specific chart?

To visualize pairwise correlations and distributions together.
