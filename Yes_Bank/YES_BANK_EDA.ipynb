### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# check if any outliers
plt.figure(figsize=(10,6))
sns.boxplot(df)
plt.tight_layout
plt.show()
````

####Here we see only few outliers but these outliers can be because of Major event ,leaving them for now based on the date will decide

````python
# Lets Check which Dates have so many ouliers
# Identify outliers using the IQR (Interquartile Range) method
def detect_outliers(df, column):
    Q1 = df[column].quantile(0.25)  # First quartile (25th percentile)
    Q3 = df[column].quantile(0.75)  # Third quartile (75th percentile)
    IQR = Q3 - Q1                   # Interquartile range
    lower_bound = Q1 - 1.5 * IQR    # Lower bound for outliers
    upper_bound = Q3 + 1.5 * IQR    # Upper bound for outliers

    # Return rows where the value is outside the bounds
    return df[(df[column] < lower_bound) | (df[column] > upper_bound)]

# Detect outliers in all price columns
outliers = pd.concat([detect_outliers(df, col) for col in ['Open', 'High', 'Low', 'Close']]).drop_duplicates()

# Plot the outlier dates on a timeline
plt.figure(figsize=(12, 6))
sns.scatterplot(data=df, x='Date', y='Close', color='skyblue', label='Normal Data')
sns.scatterplot(data=outliers, x='Date', y='Close', color='red', label='Outliers')
plt.title('Outliers in Yes Bank Stock Prices Over Time')
plt.xlabel('Date')
plt.ylabel('Close Price')
plt.legend()
plt.show()
````

#### Chart - 12

````python
# Chart - 12 visualization code
#Bollinger Bands
df['MA20'] = df['Close'].rolling(20).mean()
df['Upper_Band'] = df['MA20'] + (df['Close'].rolling(20).std() * 2)
df['Lower_Band'] = df['MA20'] - (df['Close'].rolling(20).std() * 2)

plt.figure(figsize=(12,6))
plt.plot(df['Date'], df['Close'], label='Closing Price', alpha=0.5)
plt.plot(df['Date'], df['MA20'], label='20-Day Moving Avg', color='blue')
plt.plot(df['Date'], df['Upper_Band'], label='Upper Bollinger Band', linestyle='dashed', color='red')
plt.plot(df['Date'], df['Lower_Band'], label='Lower Bollinger Band', linestyle='dashed', color='green')
plt.fill_between(df['Date'], df['Lower_Band'], df['Upper_Band'], color='gray', alpha=0.2)
plt.legend()
plt.title("Bollinger Bands for Yes Bank Closing Prices")
plt.xlabel("Year")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.show()
````

#### Chart - 13

````python
# Chart - 13 visualization code
# Calculate Rate of Change (ROC) for a 10-day period
df['ROC'] = ((df['Close'] - df['Close'].shift(10)) / df['Close'].shift(10)) * 100

# Drop NaN values generated from shifting
df = df.dropna()

# Plot ROC Chart
plt.figure(figsize=(12,6))
sns.lineplot(data=df, x='Date', y='ROC', color='blue')
plt.axhline(y=0, linestyle='dashed', color='black')  # Zero line for reference
plt.axhline(y=10, linestyle='dashed', color='green', label='Overbought (Possible Sell)')
plt.axhline(y=-10, linestyle='dashed', color='red', label='Oversold (Possible Buy)')
plt.title("Rate of Change (ROC) - Stock Price Momentum Indicator")
plt.xlabel("Year")
plt.ylabel("ROC (%)")
plt.legend()
plt.xticks(rotation=45)
plt.show()
````

##### 1. Why did you pick the specific chart?

 ROC measures the percentage change in price over 10 days.

If ROC > 10% → The stock is overbought (consider selling).

If ROC < -10% → The stock is oversold (consider buying).

Crossing the zero line → Indicates a trend reversal.

##### 2. What is/are the insight(s) found from the chart?

 Positive ROC (>10%) suggests strong upward momentum.

 Negative ROC (<-10%) suggests strong downward momentum.

 Crossing zero line means a potential trend reversal.
